```python
from google.colab import auth
import gspread
from google.auth import default
import pandas as pd

# 1. Authenticate with your Google account
print("Authenticating with Google...")
auth.authenticate_user()
creds, _ = default()
gc = gspread.authorize(creds)

# 2. Open or create the application spreadsheet
spreadsheet_name = "Application Chamber Data"
try:
    sheet = gc.open(spreadsheet_name).sheet1
    print(f"Successfully opened existing sheet: '{spreadsheet_name}'")
except gspread.SpreadsheetNotFound:
    spreadsheet = gc.create(spreadsheet_name)
    sheet = spreadsheet.sheet1
    print(f"Created new sheet: '{spreadsheet_name}'")

    # 3. Write initial operational state
    sheet.update([
        ["Timestamp", "Source", "Status", "Payload"],
        [pd.Timestamp.now().strftime("%Y-%m-%d %H:%M:%S"), "Colab-Kaggle Engine", "Active", "Initialized"]
        ])

    print("Setup complete! Data written to spreadsheet.")
```
